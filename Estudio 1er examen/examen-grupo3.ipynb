En los siguientes ejercicios utilizaremos la notación *Planning Domain Definition Language* (PDDL) sin variables. Los átomos se representarán como cadenas que no comiencen por el símbolo "-" (p.e.: Q2, P3, R3...). La negación de un átomo será un átomo con el símbolo "-" delante (p.e.: -P6, -R3,...). Un literal es ún átomo o un átomo negado. Un estado es una lista (sin duplicados) de átomos y un objetivo una lista (sin duplicados) de literales.

Codificaremos un conjunto de acciones como un diccionario de pares clave-valor donde las claves son los nombres de las acciones y los valores son tuplas de 2 tuplas. La primera tupla son los prerrequisitos y la segunda tupla los efectos.

lo codificamos como el diccionario

### Ejercicio 1 (0.75 ptos):
Escribir una función *aplicable(a,e)* que tome como entrada el nombre de una acción, un estado *e* y un conjunto de acciones en el formato especificado y devuelva *True* si la acción *a* es aplicable al estado *e*, y *False* en caso contrario.

```python
#Solución:
def aplicable(a, e, conjunto):
    prerrequisitos = conjunto[a][0]

    for p in prerrequisitos:
        if p[0] == "-":
            atomo = p[1:] # Quitamos el "-"
            if atomo in e:
                return False

        else:
            if p not in e:
                return False

    return True
```

Evaluación:

```python
aplicable("A",['P4','P5'],C1)
# Salida esperada: True
```

```python
aplicable("B",['P4','P5'],C1)
# Salida esperada: False
```

```python
aplicable("A",['P3','P1'],C1)
# Salida esperada: False
```

```python
aplicable("F",['P2','P6'],C1)
# Salida esperada: True
```

### Ejercicio 2 (0.5 pto):
Escribir una función *aplicables(e,C)* que haciendo uso de la función definida en el apartado anterior tome como entada un estado *e* y un conjunto de acciones *C* en el formato especificado y devuelva la lista de acciones de *C* que sean aplicables a *e*.

```python
#Solución:
def aplicables(e, C):
    validas = []
    for a in C:
        if aplicable(a, e, C):
            validas.append(a)

    return validas
```

### Ejercicio 3 (0.5 ptos):
A continuación, crea dos métodos: *quitar(neg,e)* y *poner(pos,e)*. El método *quitar(neg,e)* recibirá una lista de átomos *neg* y un estado *e* y deberá devolver el resultado de eliminar los átomos de *neg* del estado *e*. El método *poner(pos,e)* recibirá una lista de átomos *pos* y un estado *e* y deberá devolver el resultado de añadir los átomos de *pos* del estado *e* evitando que aparezcan átomos duplicados en *e*. 


```python
#Solución:
def quitar(neg, e):
    resultado = []

    for atomo in e:
        if atomo not in neg:
            resultado.append(atomo)
    return resultado
            
def poner(pos, e): 
    resultado = e.copy() 
    
    for atomo in pos:
        if atomo not in resultado:
            resultado.append(atomo)

    return resultado
```

```python
quitar(['R4','R5'],['R2','R4','R5','R6'])
# Salida esperada: ['R2','R6']
```

```python
poner(['Q2','Q5'],['Q1','Q2','Q3','Q4'])
# Salida esperada: ['Q1', 'Q2', 'Q3', 'Q4', 'Q5']
```

```python
poner(['R2'],['R2','R4','R5','R6'])
# Salida esperada: ['R2', 'R4', 'R5', 'R6']
```

```python
quitar(['P1','P2'],['P2'])
# Salida esperada: []
```

### Ejercicio 4 (0.5 ptos):
Haciendo uso de las funciones definidas en el apartado anterior, define la función *resultado(a,e,C)* donde *C* es un conjunto de acciones, *a* es el nombre de una de las acciones de *C* y *e* es un estado. La función debe devolver el estado resultante de aplicar la acción *a* al estado *e*. En el momento de aplicar esta función supondremos que ya hemos comprobado que *a* es aplicable a *e*, por lo que en este ejercicio no se pide comprobar la aplicabilidad. Es importante que el estado obtenido no contenga duplicados.

```python
#Solución:
def resultado(a, e, C):
    efectos = C[a][1]

    pos = []
    neg = []

    for ef in efectos:
        if ef[0] == "-":
            neg.append(ef[1:])
        else:
            pos.append(ef)

    estado_tras_borrar = quitar(neg, e)
    return poner(pos, estado_tras_borrar)
```

Evaluación:

```python
resultado('E',['P1','P2','P6'],C1)
# Salida esperada: ['P6', 'P5']
```

```python
resultado('G',['Q1','Q2'],C2)
# Salida esperada: ['Q1', 'Q2', 'R15', 'R16']
```

```python
resultado('D',['R1','R2','R4'],C3)
# Salida esperada: ['R1', 'R2', 'R4', 'Q7']
```

### Ejercicio 5 (0.5 ptos):

Vamos a definir ahora unas funciones auxiliares que nos permitirán evaluar si un determinado plan es solución a un problema de planificación dado. Define una función *cumple(estado,objetivo)*  que reciba un *estado* (una lista de átomos) y un objetivo (una lista de literales).y devuelva *True* si el estado cumple con el objetivo (posee todos los átomos positivos del *objetivo* y ninguno de sus átomos negativos).

```python
#Solución:
def cumple(estado, objetivo):
    for literal in objetivo:
        if literal[0] == "-":
            atomo = literal[1:]
            if atomo in estado:
                return False
        else:
            if literal not in estado:
                return False

    return True
```

Evaluación:

```python
estado = ("A1","A4")
objetivo = ("A1","-A2")
cumple(estado,objetivo)
# Salida esperada: True
```

```python
estado = ("A3","A4")
objetivo = ("A1","-A2")
cumple(estado,objetivo)
# Salida esperada: False
```

```python
estado = ("A1","A2")
objetivo = ("A1","-A2")
cumple(estado,objetivo)
# Salida esperada: False
```

```python
estado = ("A2","A3")
objetivo = ("A2","-A5")
cumple(estado,objetivo)
# Salida esperada: True
```

### Ejercicio 6 (0.5 ptos):
Define una función *ejecutar_plan(plan,e_inicial,C)* que reciba un plan (una lista de acciones), un estado inicial y un conjunto de acciones *C* y devuelva el estado resultante de aplicar dicho plan a partir del estado inicial.

**Nota:** En caso de que no sea posible ejecutar el plan a partir del estado incial, la función debe devolver -1.

```python
#Solución:
def ejecutar_plan(plan, e_inicial, C):
    actual = e_inicial.copy()

    for a in plan: 
        if not aplicable(a, actual, C):
            return -1

        actual = resultado(a, actual, C)

    return actual
```

Evaluación:

```python
ejecutar_plan(["F"],["P2"],C1)
# Salida esperada: ['P2', 'P6', 'P9']
```

```python
ejecutar_plan(["F","E"],["P2"],C1)
# Salida esperada: ['P6', 'P9', 'P5']
```

```python
ejecutar_plan(["F","E","C"],["P2"],C1)
# Salida esperada: -1
```

```python
ejecutar_plan(["C","E","A"],["P4","P6"],C1)
# Salida esperada: ['P4', 'P6', 'P8', 'P5', 'P7']
```

```python
ejecutar_plan(["C","E","A","E"],["P4","P6"],C1)
# Salida esperada: -1
```

### Ejercicio 7 (0.25 pto):
Define una función *es_solucion(plan,e_inicial,objetivo,C)* que recibiendo un plan (una lista de acciones), un estado inicial (una lista de átomos), un objetivo (una lista de literales) y una lista de acciones (en el formato presentado al principio del examen). La función debe devolver *True* si dicho plan permite llegar desde el estado inicial a un estado que cumpla con el objetivo y *False* en caso contrario.

```python
#Solución:
def es_solucion(plan, e_inicial, objetivo, C):
    estado_final = ejecutar_plan(plan, e_inicial, C)

    if estado_final == -1: 
        return False

    return cumple(estado_final, objetivo)
```
