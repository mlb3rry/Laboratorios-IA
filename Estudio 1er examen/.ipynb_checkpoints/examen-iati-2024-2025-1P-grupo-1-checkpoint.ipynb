En los siguientes ejercicios vamos a implementar algunas funciones relacionadas con los proceso de decisión de Markov. Definimos un proceso de decisión de Markov como una lista [R,T] donde R es un diccionario donde las claves son cadenas que representan a los estados y los valores asociados a dichas claves son los valores de las recompensas para cada uno de los estados y T es el modelo de transición. El modelo de transición es un diccionario donde las claves son pares $(s,a)$ y los valores son listas $[(s_1,p_1),...,(s_n,p_n)]$ donde $s_1,...,s_n$ son los estados a los que se puede acceder desde $s$ y $p_i$ representa la probabilidad de llegar a $s_i$ desde $s$ realizando la acción $a$. Si en el diccionario que representa al modelo de transición no existe el par $(s_i,a_j)$ significa que la acción $a_j$ no es aplicable al estado $s_i$. Algunos ejemplos son:

```python
MDP_1 = [{'s1':-0.05,'s2':-0.05,'s3':-1,'s4':1},
         {('s1','a1'):[('s1',0.1),('s2',0.8),('s3',0.1)],
          ('s1','a2'):[('s2',0.3),('s3',0.7)],
          ('s2','a1'):[('s1',0.3),('s2',0.1),('s3',0.1),('s4',0.5)],
          ('s2','a2'):[('s2',0.6),('s4',0.4)],
          ('s2','a3'):[('s1',0.9),('s4',0.1)],
          ('s2','a4'):[('s1',0.8),('s2',0.2)],
          ('s3','a1'):[('s2',0.1),('s3',0.1),('s4',0.8)],
          ('s3','a2'):[('s1',0.3),('s4',0.7)],
          ('s3','a4'):[('s1',0.6),('s2',0.1),('s3',0.1),('s4',0.2)],
          ('s4','a1'):[('s1',0.3),('s2',0.1),('s3',0.1),('s4',0.7)],
          ('s4','a2'):[('s1',0.3),('s4',0.7)],
          ('s4','a4'):[('s1',0.2),('s3',0.1),('s4',0.7)]}]

MDP_2 = [{'e1':-0.3,'e2':-0.1,'e3':2},
         {('e1','acc_1'):[('e1',0.1),('e2',0.9)],
          ('e1','acc_2'):[('e2',0.3),('e3',0.7)],
          ('e1','acc_3'):[('e1',0.3),('e2',0.1),('e3',0.6)],
          ('e2','acc_1'):[('e2',0.6),('e3',0.4)],
          ('e2','acc_2'):[('e1',1)],
          ('e3','acc_1'):[('e1',0.8),('e2',0.2)],
          ('e3','acc_3'):[('e1',0.1),('e3',0.9)],
          ('e3','acc_4'):[('e1',0.3),('e2',0.7)]}]
```

Representaremos una política como un diccionario donde las claves son los estados y los valores asociados representan las acciones recomendadas por la politica para dichos estados. 

```python
pi_1 = {'s1':'a1','s2':'a1','s3':'a1','s4':'a1'}
pi_2 = {'s1':'a2','s2':'a3','s3':'a4','s4':'a2'}

pi_3 = {'e1':'acc_2','e2':'acc_1','e3':'acc_4'}
pi_4 = {'e1':'acc_1','e2':'acc_2','e3':'acc_3'}
```

Representaremos una valoración como un diccionario en el que habrá una clave por cada estado y en donde los valores serán las valoraciones asociadas. Por ejemplo:

```python
V1 = {'s1':0,
      's2':0,
      's3':0,
      's4':0}

V2 = {'s1':0.15,
      's2':0.25,
      's3':0.35,
      's4':0.45}

V3 = {'e1':0.1,
      'e2':0.7,
      'e3':1}

V4 = {'e1':0.4,
      'e2':0.6,
      'e3':0.5}
```

**Nota**: En los siguientes ejercicios, el estudiante puede añadir tantas funciones auxiliares como considere necesario.

### Ejercicio 1 (0.5 ptos)
Define una función ep(MDP,s,pi) que tome como entrada un proceso de decisión de Markov, un estado $s$ y una política $pi$ y devuelva una lista de pares $(s',prob)$ donde los $s'$ son los estados a los que se puede llegar desde $s$ tras aplicar la acción que determina la política $pi$ y prob es la probabilidad de llegar a ese estado. Por ejemplo si tomamos el proceso de decisión MDP_1, el estado s2 y la política pi_2, vemos que la acción recomendada para s2 por esa política es a3 y que si aplicamos la acción a3 a s2 llegamos a
* s1 con probabilidad 0.9
* s4 con probabilidad 0.1

Por tanto la salida de la función debe ser [('s1', 0.9), ('s4', 0.1)]

```python
# Solución
```

```python
ep(MDP_1,'s2',pi_2)
# Respuesta esperada: [('s1', 0.9), ('s4', 0.1)]
```

```python
ep(MDP_1,'s4',pi_1)
# Respuesta esperada: [('s1', 0.3), ('s2', 0.1), ('s3', 0.1), ('s4', 0.7)]
```

```python
ep(MDP_2,'e1',pi_3)
# Respuesta esperada: [('e2', 0.3), ('e3', 0.7)]
```

```python
ep(MDP_2,'e2',pi_4)
# Respuesta esperada: [('e1', 1)]
```

### Ejercicio 2 (0.5 ptos):
Definir una función *valoracion_secuencia(MDP,sec,desc)* que reciba como dato de entrada un proceso de decisión de Markov *MDP*, una secuencia de estados *sec* y un factor de descuento *desc* y devuelva la valoración de dicha secuencia de estados: 

$$V([q_0 q_1 q_2 ...])=R(q_0)+ \gamma R(q_1) + \gamma^2 R(q_2) + ...$$

Por ejemplo, si consideramos el problema $MDP\_1$, la secuencia $s1 s2 s3 s4$ y un factor de descuento 0.5 el resultado de la valoración sería:

$$V([s1 s2 s3 s4])=R(s1)+ \gamma R(s2) + \gamma^2 R(s3) + \gamma^3 R(s4)=-0.05 + 0.5(-0.05) + 0.5^2(-1)+ 0.5^3(1)=-0.2$$


```python
# Solución
```

```python
valoracion_secuencia(MDP_1,['s1','s2','s3','s4'],0.5)
# Respuesta esperada: -0.2
```

```python
valoracion_secuencia(MDP_1,['s1','s2','s1','s2'],0.25)
# Respuesta esperada: -0.06640625
```

```python
valoracion_secuencia(MDP_2,['e1','e2','e2'],0.5)
# Respuesta esperada: -0.375
```

```python
valoracion_secuencia(MDP_2,['e2','e2','e2'],0.9)
# Respuesta esperada: -0.271
```

### Ejercicio 3 (0.75 ptos):

Definir una función *probabilidad_secuencia(MDP,sec,pi)* que reciba como dato de entrada un proceso de decisión de Markov *MDP*, una secuencia de estados *sec* y una política *pi* y devuelva la probabilidad de dicha secuencia de estados bajo la política *pi*.

Por ejemplo, si consideramos el problema $MDP\_1$, la secuencia $[s1, s2, s3, s4]$ y la política *pi_1* (que asigna la acción $a1$ a todos los estados), la probabilidad resultante sería:

$$0.8 \times 0.1 \times 0.8=0.064$$

Ya que 0.8 es la probabilidad de pasar al estado $s2$ desde el estado $s1$ aplicando la acción recomendada por la política *pi_1* para $s1$, 0.1 es la probabilidad de pasar al estado $s3$ desde el estado $s2$ aplicando la acción recomendada por la política *pi_1* para $s2$ y 0.8 es la probabilidad de pasar al estado $s4$ desde el estado $s3$ aplicando la acción recomendada por la política *pi_1* para $s3$.

**Nota**: Se recomienda el uso de la función *ep(MDP,s,pi)* definida en el Ejercicio 1.


```python
#Solución:
```

```python
probabilidad_secuencia(MDP_1,["s1","s2","s3","s4"],pi_1)
# Salida esperada: 0.06400000000000002
```

```python
probabilidad_secuencia(MDP_1,["s1","s1","s1","s1"],pi_2)
# Salida esperada: 0
```

```python
probabilidad_secuencia(MDP_2,["e1","e1","e1","e1"],pi_3)
# Salida esperada: 0
```

```python
probabilidad_secuencia(MDP_2,["e1","e1","e1","e1"],pi_4)
# Salida esperada: 0.0010000000000000002
```

    * 0.7 es a probabilidad de llegar a $s4$ desde $s4$ mediante la acción $a2$
    * 0.45 es la valoración de $s1$ en $V$

```python
# Solución
```

```python
pares(MDP_1,'s4','a2',V2)
# Respuesta esperada: [(0.3, 0.15), (0.7, 0.45)]
```

### Ejercicio 5 (0.5 ptos):

Dado un problema de decisión de Markov, una política $\pi$, una valoración inicial $V$ y un factor descuento $\gamma$, vamos a obtener una nueva valoración iterando la fórmula

$$V_{i+1}^{\pi}(s) = R(s) + \gamma \cdot \displaystyle{\mathop{\sum}_{s'}} (P(s'|s,\pi(s))\cdot V_i^{\pi}(s') )$$

Para ello, se pide definir una función nueva_V(MDP,pi,V,gamma,s) que tome como dato de entrada un problema de decisión de Markov MDP, una política pi, una valoración V, un factor de descuento gamma y un estado s. Se pide obtener un nuevo valor para el estado s aplicando la fórmula anterior tomado $V^\pi_i$ como V. Por ejemplo, si tomamos el proceso de decisión de Markov MDP_1, 0.9 como factor de descuento, la política pi_2 y el estado s4, vemos que la acción recomendada es a2 y que los estados a los que se puede llegar desde s4 por la acción de a2 son s1 con probabilidad 0.3 y s4 con probabilidad 0.7. Si consideramos como valoración V2 que valora s1 como 0.15 y s4 como 0.45 tenemos que el nuevo valor para s4 es

$$1 + 0.9 *[(0.3 * 0.15)+ (0.7 * 0.45)] = 1.324$$


```python
# Solución
```

```python
nueva_V(MDP_1,pi_2,V2,0.9,'s4')
# Salida esperada: 1.324
```

```python
nueva_V(MDP_1,pi_1,V1,0.7,'s1')
# Salida esperada: -0.05
```

```python
nueva_V(MDP_2,pi_3,V4,0.9,'e2')
# Salida esperada: 0.40400000000000014
```
